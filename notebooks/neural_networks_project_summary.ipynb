# Neural networks and simulation

Public summary notebook for a multiclass burnout-risk classification project based on multilayer perceptrons. This notebook is intentionally lightweight: it reads the stored public artifacts instead of rerunning the full training workflow.

## Repository map

- `../docs/methodology_overview.md`: experiment design and modeling choices.
- `../results/model_comparison_summary.csv`: full comparison table.
- `../results/final_model_selection.csv`: compact final report table.
- `../figures/`: curated figures used in the public report.
- `../src/burnout_multiclass_workflow.py`: clean script version of the coursework workflow.

```python
from pathlib import Path
import pandas as pd

ROOT = Path('..').resolve()
results = pd.read_csv(ROOT / 'results' / 'model_comparison_summary.csv')
final_table = pd.read_csv(ROOT / 'results' / 'final_model_selection.csv')
config = pd.read_json(ROOT / 'results' / 'experiment_config.json', typ='series')

results.head(5)[[
    'model_name',
    'epochs_trained',
    'best_val_accuracy',
    'best_val_macro_f1',
    'test_accuracy',
    'macro_f1',
]]
```

## Why this project is interesting

The task is a strong public example because the dataset is imbalanced and the notebook does not rely on accuracy alone. The final selection rule uses validation Macro F1, which is much easier to defend than choosing the best model directly on the test split.

```python
best_row = results.sort_values('best_val_macro_f1', ascending=False).iloc[0]
pd.DataFrame(
    {
        'metric': ['best_val_macro_f1', 'test_accuracy', 'macro_f1', 'Low_f1', 'Medium_f1', 'High_f1'],
        'value': [
            best_row['best_val_macro_f1'],
            best_row['test_accuracy'],
            best_row['macro_f1'],
            best_row['Low_f1'],
            best_row['Medium_f1'],
            best_row['High_f1'],
        ],
    }
)
```

## Figure gallery

These are the four figures used to tell the final public story: the imbalance context, the learning behaviour of the selected model, the model-comparison view, and the final confusion matrix.

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

figure_specs = [
    ('class_distribution_and_weights.png', 'Class distribution and class weights'),
    ('training_history_reference.png', 'Best-model learning curves'),
    ('model_comparison_summary.png', 'Model comparison summary'),
    ('final_confusion_matrix.png', 'Final confusion matrix'),
]

fig, axes = plt.subplots(2, 2, figsize=(16, 12))
for ax, (filename, title) in zip(axes.flat, figure_specs):
    ax.imshow(mpimg.imread(ROOT / 'figures' / filename))
    ax.set_title(title)
    ax.axis('off')

plt.tight_layout()
```
